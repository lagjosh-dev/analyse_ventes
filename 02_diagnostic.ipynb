```python
# Calcul des écart sur ventes

df = donnees_test["Sales 2020 T1.csv"].copy()

# Calcul du montant théorique
df["ventes_calculees"] = (
    df["quantite_vente"] * df["prix_unitaire"]
    - df["montant_remise"]
)

# Calcul de l'écart
df["ecart_ventes"] = (
    df["montant_ventes"] - df["ventes_calculees"]
)

# Afficher les résultats
display(df[[
    "id_vente",
    "quantite_vente",
    "prix_unitaire",
    "montant_remise",
    "montant_ventes",
    "ventes_calculees",
    "ecart_ventes"
]].head(10))
```

```python
# Identifier les écarts supérieurs à 0,01 $
anomalies_ventes = df[df["ecart_ventes"].abs() > 0.01]

# Résumer les résultats
print("Nombre total de ventes :", len(df))
print("Nombre d'écarts détectés :", len(anomalies_ventes))
print("Pourcentage d'écarts :", round(len(anomalies_ventes) / len(df) * 100, 2), "%")

# Afficher les ventes présentant des écarts
display(anomalies_ventes[[
    "id_vente",
    "montant_ventes",
    "ventes_calculees",
    "ecart_ventes"
]].head(10))
```

Les résultats ne permettent pas encore de conclure à l'existence de 61 anomalies comptables.

Décision : arrondir les écarts à deux décimales, puis renouveler le contrôle afin de distinguer les différences de précision numérique des écarts de calcul significatifs.

Aucune modification des données originales n'est effectuée à cette étape.

```python
# Arrondir les écarts à deux décimales
df["ecart_ventes"] = df["ecart_ventes"].round(2)

# Identifier les écarts supérieurs à un centime
anomalies_ventes = df[df["ecart_ventes"].abs() > 0.01]

print("Nombre d'écarts :", len(anomalies_ventes))

display(anomalies_ventes[[
    "id_vente",
    "montant_ventes",
    "ventes_calculees",
    "ecart_ventes"
]].head(10))
```
